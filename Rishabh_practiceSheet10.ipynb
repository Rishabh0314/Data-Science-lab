24. Calculate each product's total cost, the overall bill, and each product's percentage contribution to the bill.
quantity = np.array([3, 5, 2, 7])
price = np.array([120, 80, 250, 60])

```python
quantity = np.array([3, 5, 2, 7])
price = np.array([120, 80, 250, 60])
total_cost = quantity * price
overall_bill = np.sum(total_cost)
product_percentages = (total_cost / overall_bill) * 100
print("Total Cost:", total_cost)
print("Overall Bill:", overall_bill)
print("Product Percentages:", product_percentages)
```
