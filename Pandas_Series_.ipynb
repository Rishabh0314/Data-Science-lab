```python
#question 29
import pandas as pd

sales = pd.Series(
    [300, 750, 450, 900],
    index=['Pen', 'Notebook', 'Bag', 'Laptop']
)

print("Values:")
print(sales.values)

print("\nIndex:")
print(sales.index)

print("\nNotebook sales:")
print(sales['Notebook'])

print("\nSales above 500:")
print(sales[sales > 500])

print("\nIs Bag present?")
print('Bag' in sales.index)

print("\nDictionary:")
print(sales.to_dict())
```

```python
#question 30
import pandas as pd

data = {'A': 15, 'B': 35, 'C': 25}

s = pd.Series(
    data,
    index=['C', 'A', 'D', 'B']
)

s.name = 'Observations'
s.index.name = 'Category'

print(s)
print("\nFiltered:")
print(s[s > 20])
print("\nNot null:")
print(pd.notnull(s))
print("\nSeries name:", s.name)
print("Index name:", s.index.name)
```
