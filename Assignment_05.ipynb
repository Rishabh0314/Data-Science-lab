#question 19
Modifying a dictionary directly while iterating over it changes its size during iteration and can cause a RuntimeError

```python
data = {'A': 10, 'B': 20, 'C': 30}

remove_keys = []

for key in data:
    if data[key] > 15:
        remove_keys.append(key)

for key in remove_keys:
    data.pop(key)

print(data)
```

```python
#question 20
student = {'name': 'Alex', 'age': 20, 'city': 'Delhi'}

for key, value in student.items():
    print(key, value)
```

6. Sets

#question 26
1. A set contains unique elements.
2. A set is unordered/unindexed.
3. A set is mutable, so elements can be added or removed.

```python
#question 27
colors = set(['blue', 'green', 'red'])

colors.add('blue')

print(colors)
```

The set remains unchanged because sets do not allow duplicate elements

#question 28
.discard() does not raise an error if the element is missing.
.remove() raises a KeyError if the element is missing.

```python
colors = {'red', 'blue'}

colors.discard('green')   # No error
```

```python
colors.remove('green')    # KeyError
```
